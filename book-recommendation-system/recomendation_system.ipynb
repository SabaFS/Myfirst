```python
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.metrics.pairwise import cosine_similarity
df['genres'] = df['genres'].apply(lambda x: eval(x) if isinstance(x, str) else x)

# One-hot encode the genres
mlb = MultiLabelBinarizer()
genres_encoded = mlb.fit_transform(df['genres'])

# Create a DataFrame for the one-hot encoded genres
genres_df = pd.DataFrame(genres_encoded, columns=mlb.classes_, index=df.index)

# Concatenate the genres back to the main DataFrame
df = pd.concat([df, genres_df], axis=1)

# Add bookId column for indexing (this could just be the row index or any identifier)
df['bookId'] = df.index
```

```python
# Example: Select a sample of 10,000 books
# Sample a subset of the books (ensure the index is preserved)
df_sampled = df.sample(10000, random_state=42).reset_index(drop=True)


genres_encoded_sampled = mlb.fit_transform(df_sampled['genres'])

# Compute the cosine similarity for the sampled data
similarity_matrix_sampled = cosine_similarity(genres_encoded_sampled)
```

```python
import pandas as pd
import numpy as np

def hybrid_recommend_by_genre(genre, books_df, similarity_matrix_sampled, top_n=10):
    """
    Recommends books based on a given genre by calculating the similarity between books 
    and combining that with their ratings.
    
    Parameters:
    genre (str): The genre to search for.
    books_df (pd.DataFrame): DataFrame containing book details (title, author, genres, rating, etc.).
    similarity_matrix_sampled (np.array): Cosine similarity matrix based on genres.
    top_n (int): Number of top recommendations to return.
    
    Returns:
    pd.DataFrame: DataFrame containing the top recommended books.
    """
    # Find the indices of books that match the given genre
    genre_book_indices = books_df[books_df['genres'].str.contains(genre, case=False)].index.tolist()
    
    # Debugging: Output the number of books found for the specified genre
    print(f"Books found for genre '{genre}': {len(genre_book_indices)}")
    
    # Pre-compute the ratings to avoid repeated access
    ratings = books_df['rating'].values
    
    # A dictionary to hold the cumulative scores for each book
    recommendations = {}

    for idx in genre_book_indices:
        if idx >= len(similarity_matrix_sampled):
            print(f"Index {idx} is out of bounds for the similarity matrix.")  # Debugging line
            continue
        
        # Get the cosine similarity scores for this book with all other books
        similarity_scores = similarity_matrix_sampled[idx]
        
        # Weighted scores calculation (similarity * rating)
        weighted_scores = similarity_scores * ratings
        
        # Update the recommendations dictionary with the cumulative scores
        for i, score in enumerate(weighted_scores):
            if i in recommendations:
                recommendations[i] += score
            else:
                recommendations[i] = score
    
    # Convert the recommendations dictionary to a DataFrame
    recommendations_df = pd.DataFrame(recommendations.items(), columns=['book_index', 'weighted_score'])
    
    # Merge with the original books DataFrame to get details
    recommendations_df = recommendations_df.merge(books_df, left_on='book_index', right_index=True)

    # Sort by the weighted score and get the top N recommendations
    recommendations_df = recommendations_df.sort_values(by='weighted_score', ascending=False)

    # Debugging: Output the total recommendations found
    print(f"Total recommendations found: {len(recommendations_df)}")
    
    return recommendations_df[['title', 'author', 'genres', 'rating', 'weighted_score']].head(top_n)
```
